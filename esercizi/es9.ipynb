Exercise 9.3

Write a program that plots the exponential probability distribution with a fixed parameter t0.


```python
from likelihood import exp_pdf
import numpy as np
import matplotlib.pyplot as plt
tau = 2
fig, ax = plt.subplots()
ax.set_title('Exponential PDF', size = 14)
ax.set_ylabel('pdf(x)', size = 12)
ax.set_xlabel('time', size = 12)
x_axis = np.linspace( 0., 4., 10000)
y_axis = np.arange( 0., x_axis.size)
for i in range(x_axis.size):
    y_axis[i] = exp_pdf( x_axis[i] , tau) 
plt.plot(x_axis, y_axis, color = 'green')
plt.show()
```
